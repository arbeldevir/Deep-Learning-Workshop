```python
!pip install -q pandas openpyxl

import pandas as pd

# List of exact file names
file_paths = [
    '_aligned_analysis_הודיה.xlsx',
    'aligned_analysis_זאב_complete_all_referents.xlsx',
    'LENA_MERGED.xlsx',
    'MERAV_MERGED.xlsx'
]

# Standardize column name options
possible_sentence_cols = ['full_raw_response', 'Full_Raw_Sentence']

file_stats = []
all_sentences = set()

for file in file_paths:
    try:
        df = pd.read_excel(file)

        # Identify the relevant sentence column
        sentence_col = next((col for col in df.columns if col in possible_sentence_cols), None)

        if sentence_col is None:
            print(f"⚠️ Could not find sentence column in '{file}'. Found columns: {list(df.columns)}")
            continue

        # Total rows / lines (meaning values) in this file
        total_lines = len(df)

        # Unique sentences in this file
        unique_sentences_in_file = set(df[sentence_col].dropna().astype(str).str.strip())
        unique_sentence_count = len(unique_sentences_in_file)

        # Track overall unique sentences across all files
        all_sentences.update(unique_sentences_in_file)

        file_stats.append({
            'file': file,
            'total_lines': total_lines,
            'unique_sentences': unique_sentence_count,
            'sentences_set': unique_sentences_in_file
        })

    except FileNotFoundError:
        print(f"❌ File not found: '{file}'. Ensure it is uploaded to Colab.")

# Global total of unique sentences across all combined files
total_unique_sentences = len(all_sentences)
total_lines_all_files = sum(item['total_lines'] for item in file_stats)

results = []
for item in file_stats:
    coverage_pct = (item['unique_sentences'] / total_unique_sentences * 100) if total_unique_sentences > 0 else 0
    avg_dissections_per_sentence = (item['total_lines'] / item['unique_sentences']) if item['unique_sentences'] > 0 else 0

    results.append({
        'File Name': item['file'],
        'Total Lines (Meaning Values)': item['total_lines'],
        'Unique Sentences': item['unique_sentences'],
        'Sentence Coverage (%)': f"{coverage_pct:.2f}%",
        'Avg Lines / Sentence': f"{avg_dissections_per_sentence:.2f}"
    })

summary_df = pd.DataFrame(results)

print("=== PER-FILE SUMMARY ===")
display(summary_df)

print("\n=== COMBINED OVERALL TOTALS ===")
print(f"Total Meaning Value Lines Across All Files: {total_lines_all_files}")
print(f"Total Unique Raw Sentences Across All Files: {total_unique_sentences}")
```
